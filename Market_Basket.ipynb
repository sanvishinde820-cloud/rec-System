```python
# ===============================
# IMPORT LIBRARIES
# ===============================
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# ===============================
# LOAD CLEANED DATA
# ===============================
df = pd.read_csv("cleaned_retail.csv")

# ===============================
# KEEP TOP 20 PRODUCTS
# ===============================
top_products = df['Description'].value_counts().head(20).index
df = df[df['Description'].isin(top_products)]

# ===============================
# CREATE BASKET MATRIX
# ===============================
basket = (df.groupby(['InvoiceNo', 'Description'])['Quantity']
            .sum()
            .unstack()
            .fillna(0)
            .set_index(df.groupby('InvoiceNo').size().index))

# Convert to 0/1
basket = basket.applymap(lambda x: 1 if x > 0 else 0)

# ===============================
# APRIORI
# ===============================
frequent_itemsets = apriori(basket, 
                            min_support=0.005, 
                            use_colnames=True,
                            max_len=2)

# ===============================
# ASSOCIATION RULES
# ===============================
rules = association_rules(frequent_itemsets,
                          metric="confidence",
                          min_threshold=0.3)
results = results.sort_values(by="confidence", ascending=False)
results = results.head(3)


rules = rules.sort_values(by='lift', ascending=False)

print(rules.head(10))

print(len(df))



```
